```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
import pandas as pd
```

```python
col_names=['Pregnant','Glucose','BP','BMI','Skin','Insulin','Pedigree','Age','Label']
pima=pd.read_csv("diabetes.csv",comment='#',names=col_names,header=0)
pima.head()
```

```python
pima.columns
```

```python
x=pima[['Pregnant', 'Glucose', 'BP', 'BMI', 'Skin', 'Insulin', 'Pedigree',
       'Age']]
y=pima['Label']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=43)
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
model=LogisticRegression(max_iter=500,random_state=1)
model.fit(x_train_scaled,y_train)
y_pred=model.predict(x_test_scaled)
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report,ConfusionMatrixDisplay
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(ConfusionMatrixDisplay.from_predictions(y_test,y_pred))
```
